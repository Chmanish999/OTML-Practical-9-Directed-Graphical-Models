# OTML Practical 9: Directed Graphical Models

Directed Graphical Models represent relationships among variables using a directed graph.

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.DiGraph()

G.add_node("Age")
G.add_node("Gender")
G.add_node("Music Genre")

G.add_edge("Age", "Music Genre")
G.add_edge("Gender", "Music Genre")
```

```python
plt.figure(figsize=(7, 5))

pos = nx.spring_layout(G, seed=42)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=3000,
    font_size=12,
    arrows=True
)

plt.title("Directed Graphical Model for Music Recommendation")
plt.show()
```

```python
print("Dependencies in the Directed Graphical Model:")

for edge in G.edges():
    print(edge[0], "influences", edge[1])
```

```python
print("""
Interpretation:

1. Age is a parent node of Music Genre.
2. Gender is a parent node of Music Genre.
3. Music Genre depends on Age and Gender.
4. This structure represents a simple Directed Graphical Model.
""")
```
